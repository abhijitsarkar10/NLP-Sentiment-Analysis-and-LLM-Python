#### Type1

```python
from textblob import TextBlob

# Define a list of sample texts
texts = [
    "I love this movie, it's fantastic!",
    "This restaurant is terrible, I wouldn't recommend it to anyone.",
    "The weather today is just perfect.",
    "I feel indifferent about this book.",
    "There is a terrorist attack happened in Mumbai.",
    "The customer service was excellent, I'm very satisfied.",
    "Smoking is injurious to health"
]

# Analyze sentiment for each text
for text in texts:
    sentiment_score, subjectivity_score = TextBlob(text).sentiment
    if sentiment_score > 0:
        sentiment_label = "positive"
    elif sentiment_score < 0:
        sentiment_label = "negative"
    else:
        sentiment_label = "neutral"
    print(f"Text: {text}")
    print(f"Sentiment: {sentiment_label}")
    print(f"Sentiment Score: {sentiment_score}")
    print(f"Subjectivity Score: {subjectivity_score}")
"""
Subjective Score:
0: Indicates highly objective text, typically consisting of factual statements, descriptions, or observations that are not influenced by personal feelings or opinions.
1: Indicates highly subjective text, containing personal opinions, emotions, or interpretations.
"""
```

```python
# Trained on IMDB ratings that's why this will only give binary response positive/Negative
from flair.models import TextClassifier
from flair.data import Sentence

classifier = TextClassifier.load('sentiment')

texts = [
    "I don't like this movie, it's fantastic!",
    "This restaurant is terrible, I wouldn't recommend it to anyone.",
    "The weather today is just perfect.",
    "I feel indifferent about this book.",
    "There is a terrorist attack happened in Mumbai.",
    "The customer service was excellent, I'm very satisfied.",
    "Smoking is injurious to health"
]

for text in texts:
    sentence = Sentence(text)
    classifier.predict(sentence)
    print(f"Text: {text}")
    print(f"Sentiment: {sentence.labels[0]}")
```

#### Type3

```python
# Great for tweets, reviews, or short English sentences.
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import nltk
nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()

texts = [
    "I love this movie, it's fantastic!",
    "This restaurant is terrible, I wouldn't recommend it to anyone.",
    "The weather today is just perfect.",
    "I feel indifferent about this book.",
    "There is a terrorist attack happened in Mumbai.",
    "The customer service was excellent, I'm very satisfied."
]

for text in texts:
    score = sia.polarity_scores(text)
    compound = score['compound']
    sentiment = 'positive' if compound > 0.05 else 'negative' if compound < -0.05 else 'neutral'
    print(f"Text: {text}")
    print(f"Sentiment: {sentiment}, Compound Score: {compound}")
```

```python
import os
from langchain_community.llms import CTransformers
from langchain_core.callbacks.streaming_stdout import StreamingStdOutCallbackHandler
```

```python
model_id = 'TheBloke/Orca-2-13B-GGUF'
```

```python
os.environ['XDG_CACHE_HOME'] = 'drive/MyDrive/LLM_data/model/cache/'
config = {'temperature':0.9,'max_new_tokens': 512, 'context_length':4000,'gpu_layers':50,'repetition_penalty':1 }
llm = CTransformers(model=model_id,
                    model_type="llama",
                    gpu_layers=50,
                    device = 0,
                    config=config,
                    callbacks=[StreamingStdOutCallbackHandler()],
                   )
```

```python
llm.invoke('write a short story')
```

```python
llm.invoke("what is cupcake?")
```
